## 🔍 Step 1: Exploration
In this step, we explore the dataset to understand its structure and contents.  
We use functions like `.head()`, `.info()`, and `.describe()` to view the data, check data types, and see basic statistics.


```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

## 🧹 Step 2: Data Cleaning
In this step, we clean the dataset by handling missing values and unnecessary data.  
- Fill missing values in **Age** using the median  
- Fill missing values in **Embarked** using the mode  
- Drop the **Cabin** column due to too many missing values  
- Remove duplicate rows if any  


```python
missing_age_count = df['Age'].isna().sum()
print(f"Number of missing Age values: {missing_age_count}")
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
print("Missing Age after filling:", df['Age'].isna().sum())
```

```python
# Check missing values in Embarked
print("Missing Embarked values:", df['Embarked'].isna().sum())

# Fill missing values with the mode (most common port)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Verify
print("Missing Embarked after filling:", df['Embarked'].isna().sum())
```

```python
# Drop Cabin safely
df.drop(columns=['Cabin'], errors='ignore', inplace=True)
```

```python
print(df.columns)
```

```python
# Check for duplicate rows
num_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {num_duplicates}")
```

```python
if num_duplicates > 0:
    df.drop_duplicates(inplace=True)
    print("Duplicates removed.")
else:
    print("No duplicates found.")
```

## 📊 Step 3: Data Analysis
In this step, we analyze the data using `groupby()` to find patterns and relationships.  
We calculate survival rates by gender, class, and age group, and also find the average age per class.


```python
#Survival rate by gender
survival_by_gender = df.groupby('Sex')['Survived'].mean()
print("Survival rate by gender:\n", survival_by_gender)
```

```python
#Survival rate by class
survival_by_class = df.groupby('Pclass')['Survived'].mean()
print("\nSurvival rate by class:\n", survival_by_class)
```

```python
#Average age per class
avg_age_per_class = df.groupby('Pclass')['Age'].mean()
print("\nAverage age per class:\n", avg_age_per_class)
```

```python
#Survival rate by age group
df['AgeGroup'] = pd.cut(df['Age'], bins=[0,12,18,35,60,100],
                        labels=['Child','Teen','Young Adult','Adult','Senior'])

survival_by_agegroup = df.groupby('AgeGroup')['Survived'].mean()
print("\nSurvival rate by age group:\n", survival_by_agegroup)
```

## 🔍 Step 4: Filtering
In this step, we filter the dataset to focus on specific groups of passengers.  
We extract female survivors, children who survived, and first-class passengers with high survival probability.

```python
#Female passengers who survived
female_survived = df[(df['Sex'] == 'female') & (df['Survived'] == 1)]
print("Number of female passengers who survived:", len(female_survived))
```

```python
#Children who survived
children_survived = df[(df['Age'] < 18) & (df['Survived'] == 1)]
print("Number of children who survived:", len(children_survived))
```

```python
#1st class passengers with high survival probability
first_class_survived = df[(df['Pclass'] == 1) & (df['Survived'] == 1)]
print("Number of 1st class passengers who survived:", len(first_class_survived))
```
